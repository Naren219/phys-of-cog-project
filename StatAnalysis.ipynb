```python
from pathlib import Path
import mne
import utils
import numpy as np
from sklearn.decomposition import FastICA
import matplotlib.pyplot as plt

# Import helpers from utils.py (same folder as this notebook)
from utils import (
    load_eeg_data,
    plot_all_channels_by_trial,
    plot_all_trials_by_channel,
)

cwd = Path.cwd()
dataset_path = cwd / "dataset"

print("Current folder:", cwd)
print("Imported utils from:", Path(utils.__file__).resolve())
print("Expected dataset folder:", dataset_path)
print("Dataset exists:", dataset_path.exists())

if not dataset_path.exists():
    print("Tip: run this notebook from Proyect_EEG/ and create Proyect_EEG/dataset/")

ip = get_ipython()
if ip is not None:
    ip.run_line_magic("matplotlib", "qt")
    mne.viz.set_browser_backend("qt")
print("MNE browser backend:", mne.viz.get_browser_backend())
```

```python
conditions = ["BLA", "BLT", "P1", "P2", "P3"]
subjectID = 5
```

```python
from mne.preprocessing import ICA
ica = ICA(n_components=0.99, method='fastica', random_state=97)
powerDict = {}
maxMean = {} # Condition : (maximum mean power, ICA component)
ica_dict = {}

for condition in conditions:
    epochs = load_eeg_data(condition=condition, subject_id=subjectID, trials="all", channels="all") 
    epochs.load_data()
    epochs.filter(1., None)
    ica = ICA(n_components=0.99, method='fastica', random_state=97)
    ica.fit(epochs)
    epochs_clean = ica.apply(epochs.copy())
    sources = ica.get_sources(epochs_clean).get_data()
    power = np.var(sources, axis=2)
    mean_power = power.mean(axis=0)
    powerDict[condition] = mean_power
    max_idx = np.argmax(mean_power)
    max_val = mean_power[max_idx]
    maxMean[condition] = (max_val, max_idx)
    ica_dict[condition] = ica

n_cond = len(maxMean)
fig, axes = plt.subplots(1, n_cond, figsize=(4*n_cond, 4))

if n_cond == 1:
    axes = [axes]

for ax, cond in zip(axes, maxMean.keys()):
    max_val, comp_idx = maxMean[cond]
    ica = ica_dict[cond]
    component_map = ica.get_components()[:, comp_idx]
    mne.viz.plot_topomap(component_map, ica.info, axes=ax, show=False)
    ax.set_title(f"{cond}\nComp {comp_idx}\nPower={max_val:.2e}")

fig, ax = plt.subplots()
for cond, condMeanPower in powerDict.items():
    ax.plot(condMeanPower, marker='o', linestyle='-', label=cond)

ax.set_xlabel("ICA Component")
ax.set_ylabel("Mean Power")
ax.set_title("Mean ICA Power per Component")
ax.legend()  
plt.show()
```
